```python
# NP41. Create a synthetic ML dataset using NumPy.
# 💡 **Hint:** Use `np.random.normal`, `np.random.randint`, `np.random.choice` to generate features. Use `np.where` or `np.select` to create a binary/multi-class label column.

##**Answer:**

np.random.seed(42)

df = pd.DataFrame({
    "Age": np.random.randint(20, 60, 10),
    "Study_Hours": np.random.normal(5, 2, 10).round(2),
    "Department": np.random.choice(["DS", "ML", "AI"], 10)
})

df["Study_Hours"] = np.clip(df["Study_Hours"], 0, None)

df["Result"] = np.where(df["Study_Hours"] >= 5, 1, 0)

print(df)
 

##**Explanation:**
##-    `np.random.randint(20, 60, 10)` generates 10 ages from 20 to 59.
##-    `np.random.normal(5, 2, 10)` generates 10 values around a mean of 5.
##-    `np.random.choice()` selects random department names.
##-    `np.where()` creates the binary target: 1 if Study_Hours ≥ 5, otherwise 0.
##-    `np.clip()` prevents negative study hours.


### NP42. Detect and replace outliers using NumPy.
#> 💡 **Hint:** Compute Q1, Q3 with `np.percentile`. Define bounds as `Q1 #-    1.5*IQR` and `Q3 + 1.5*IQR`. Use `np.clip(arr, lo, hi)` to replace outliers with boundary values.

##**Answer:**

 
df = pd.DataFrame({
    "Salary": [25000, 30000, 28000, 35000, 40000, 200000]
})

arr = df["Salary"].to_numpy()

Q1, Q3 = np.percentile(arr, [25, 75])
IQR = Q3 #-    Q1

lo = Q1 #-    1.5 * IQR
hi = Q3 + 1.5 * IQR

print("Lower Bound:", lo)
print("Upper Bound:", hi)

df["Salary_Cleaned"] = np.clip(arr, lo, hi)

print(df)


##**Explanation:**
#1. Calculate Q1 and Q3.
#2. Calculate `IQR = Q3 #-    Q1`.
#3. Calculate the lower and upper boundaries.
#4. `np.clip()` replaces values outside those boundaries with the nearest boundary.

#**Note:** This method caps outliers; it does not remove rows containing them.


### NP43. Scale dataset using NumPy functions.
#> 💡 **Hint:** Apply Min-Max or Z-score scaling using NumPy on the `.values` array. Store results as new columns in the DataFrame.
df = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60],
    "Salary": [10000, 20000, 30000, 40000, 50000]
})

arr = df[["Age", "Salary"]].to_numpy(dtype=float)

min_values = arr.min(axis=0)
max_values = arr.max(axis=0)

denominator = max_values #-    min_values
denominator[denominator == 0] = 1

scaled = (arr -    min_values) / denominator

df[["Age_Scaled", "Salary_Scaled"]] = scaled

print(df)
 
##**Explanation:**

#The formula is:

#`(value #-    minimum) / (maximum #-    minimum)`

##-    Minimum value becomes 0.
##-    Maximum value becomes 1.
##-    Other values fall between 0 and 1.
##-    `axis=0` calculates the minimum and maximum separately for each column.


### NP44. Engineer new features using NumPy.
#> 💡 **Hint:** Create polynomial features (`np.square`, `np.power`), log features (`np.log1p`), interaction features (column1 × column2), and binned features (`np.digitize`).
 
df = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60],
    "Salary": [10000, 20000, 30000, 40000, 50000]
})

df["Age_Squared"] = np.square(df["Age"].to_numpy())
df["Age_Cubed"] = np.power(df["Age"].to_numpy(), 3)

df["Salary_Log"] = np.log1p(df["Salary"].to_numpy())

df["Age_Salary_Interaction"] = (
    df["Age"].to_numpy() * df["Salary"].to_numpy()
)

bins = [30, 45]
df["Age_Group"] = np.digitize(df["Age"].to_numpy(), bins)

print(df)
##**Explanation:**
##-    `np.square()` creates Age².
##-    `np.power()` creates Age³.
##-    `np.log1p()` calculates `log(1 + Salary)`, which is useful for reducing the effect of large positive values.
##-    Multiplying Age by Salary creates an interaction feature.
##-    `np.digitize()` assigns a numeric bin based on the boundaries: values ≤30 get 0, values >30 and ≤45 get 1, and values >45 get 2.

### NP45. Use exponential functions (`np.exp`).
#> 💡 **Hint:** `np.exp(df["col"])` computes `e^x` for each element. This is the inverse of `np.log`. Useful for reverting log-transformed values.

##**Answer:**

 
df = pd.DataFrame({
    "Salary": [10000, 20000, 30000, 40000, 50000]
})

df["Salary_Log"] = np.log(df["Salary"].to_numpy())

df["Salary_Recovered"] = np.exp(df["Salary_Log"])

print(df)
 

##**Explanation:**
##-    `np.log(x)` calculates the natural logarithm.
#-    `np.exp(x)` calculates \(e^x\).
#-    Therefore, `np.exp(np.log(x))` returns approximately the original positive value `x`.

#For values transformed using `np.log1p()`, use `np.expm1()` to reverse the transformation.


### NP46. Compute differences using `np.diff`.
#> 💡 **Hint:** `np.diff(df["col"].values)` returns an array of consecutive differences (length = n−1). Prepend `np.nan` and assign back: `df["col_diff"] = np.insert(np.diff(arr), 0, np.nan)`.

#**Answer:**

import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Sales": [100, 150, 130, 200, 250]
})

arr = df["Sales"].to_numpy()

df["Sales_Difference"] = np.insert(
    np.diff(arr), 0, np.nan
)

print(df)
 

#**Explanation:**
#-    150 − 100 = 50
#-    130 − 150 = -20
#-    200 − 130 = 70
#-    250 − 200 = 50

#`np.diff()` calculates the difference between consecutive values. The first row has no previous value, so its difference is `NaN`.


### NP47. Apply conditional encoding using NumPy.
#> 💡 **Hint:** Use `np.select` with a list of conditions and corresponding integer codes to encode categorical columns numerically. For example: `DS → 0`, `ML → 1`, `AI → 2`.

##**Answer:**

 
df = pd.DataFrame({
    "Department": ["DS", "ML", "AI", "DS", "AI"]
})

conditions = [
    df["Department"].to_numpy() == "DS",
    df["Department"].to_numpy() == "ML",
    df["Department"].to_numpy() == "AI"
]

choices = [0, 1, 2]

df["Department_Code"] = np.select(
    conditions,
    choices,
    default=-1
)

print(df)
 

#**Explanation:**
#-    DS becomes 0.
#-    ML becomes 1.
#-    AI becomes 2.
#-    Unknown categories become -1.

#This encoding assigns numbers to categories; it does not mean AI is greater than ML or ML is greater than DS.


### NP48. Build an ML-ready dataset (feature matrix X and label vector y).
#> 💡 **Hint:** Select numeric feature columns → call `.to_numpy(dtype=np.float32)` → store as `X`. Create a binary label: `y = (df["col"] >= threshold).astype(int).to_numpy()`. Check shapes with `X.shape` and `y.shape`.

##**Answer:**

 
df = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40],
    "Study_Hours": [2, 4, 5, 7, 8],
    "Score": [45, 55, 65, 80, 90]
})

X = df[["Age", "Study_Hours"]].to_numpy(
    dtype=np.float32
)

y = (df["Score"] >= 60).astype(int).to_numpy()

print("Feature Matrix X:")
print(X)

print("Label Vector y:")
print(y)

print("X shape:", X.shape)
print("y shape:", y.shape)
 

#**Explanation:**
#- `X` contains the input features Age and Study_Hours.
#- y` contains the target labels: 1 if Score ≥ 60, otherwise 0.
#- `X.shape` is `(5, 2)`: five rows and two features.
#- `y.shape` is `(5,)`: five target labels.

#For an actual ML model, split the data into training and testing sets before fitting and evaluating the model.

### NP49. Perform a mini data analysis project.
#> 💡 **Hint:** Follow these steps: (1) Generate data with NumPy, (2) Store in DataFrame, (3) Clean with Pandas, (4) Engineer features with NumPy, (5) GroupBy analysis, (6) Detect outliers, (7) Export summary.

#**Answer:**

 
np.random.seed(42)

df = pd.DataFrame({
    "Department": np.random.choice(
        ["IT", "HR", "Finance"], 20
    ),
    "Experience": np.random.randint(1, 16, 20),
    "Salary": np.random.randint(20000, 80001, 20)
})

# 1. Inspect data
print(df.head())
print(df.isnull().sum())

# 2. Clean data
df = df.drop_duplicates().copy()
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# 3. Engineer a new feature
df["Annual_Salary"] = np.multiply(
    df["Salary"].to_numpy(), 12
)

# 4. GroupBy analysis
summary = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Average_Experience=("Experience", "mean"),
    Employee_Count=("Salary", "count")
)

print(summary)

# 5. Detect salary outliers using IQR
arr = df["Salary"].to_numpy()

Q1, Q3 = np.percentile(arr, [25, 75])
IQR = Q3 #-    Q1

lower = Q1 #-    1.5 * IQR
upper = Q3 + 1.5 * IQR

df["Is_Outlier"] = (arr < lower) | (arr > upper)

print(df[df["Is_Outlier"]])

# 6. Export summary
summary.to_csv("department_summary.csv")

print("Summary exported successfully.")
 

#**Explanation:**
#This mini-project generates data, inspects and cleans it, creates Annual_Salary, calculates department statistics, identifies salary outliers, and exports the summary to a CSV file.

#---

### NP50. Build an end-to-end pipeline using NumPy + Pandas.
#> 💡 **Hint:** Chain all steps together:  
#> > 1. `np.random.seed(X)` → generate raw data  
#> > 2. `np.clip` → clean invalid values  
#> > 3. `np.log`, `np.square`, `np.where` → feature engineering  
#> > 4. `df.groupby().agg(...)` → statistical analysis  
#> > 5. IQR + `np.clip` → outlier removal  
#> > 6. `df[features].to_numpy(dtype=np.float32)` → ML feature matrix  
#> > 7. Export with `df.to_csv(...)`

#**Answer:**

 
import numpy as np
import pandas as pd

# 1. Generate raw data
np.random.seed(42)

df = pd.DataFrame({
    "Age": np.random.randint(20, 61, 30),
    "Salary": np.random.normal(40000, 12000, 30),
    "Department": np.random.choice(
        ["IT", "HR", "Finance"], 30
    )
})

# 2. Clean invalid values
df["Salary"] = np.clip(
    df["Salary"].to_numpy(), 1, None
)

# 3. Feature engineering
df["Salary_Log"] = np.log(df["Salary"].to_numpy())
df["Age_Squared"] = np.square(df["Age"].to_numpy())

df["High_Salary"] = np.where(
    df["Salary"] >= 40000, 1, 0
)

# 4. Statistical analysis
summary = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Employee_Count=("Salary", "count")
)

print(summary)

# 5. Detect and cap salary outliers
arr = df["Salary"].to_numpy()

Q1, Q3 = np.percentile(arr, [25, 75])
IQR = Q3 #-    Q1

lower = Q1 #-    1.5 * IQR
upper = Q3 + 1.5 * IQR

df["Salary_Capped"] = np.clip(arr, lower, upper)

# 6. Create ML feature matrix X and label vector y
features = [
    "Age",
    "Salary_Capped",
    "Salary_Log",
    "Age_Squared"
]

X = df[features].to_numpy(dtype=np.float32)
y = df["High_Salary"].to_numpy(dtype=np.int32)

print("X shape:", X.shape)
print("y shape:", y.shape)

# 7. Export data and summary
df.to_csv("processed_dataset.csv", index=False)
summary.to_csv("department_summary.csv")

print("Pipeline completed.")
 

#**Explanation:**
#1. NumPy generates the raw data.
#2. `np.clip()` keeps Salary positive.
#3. NumPy creates log, squared, and binary features.
#4. Pandas calculates department-level statistics.
#5. IQR boundaries identify the range used to cap outliers.
#6. `X` contains the model inputs; `y` contains the target.
#7. Both the processed dataset and summary are exported.

#---
```
